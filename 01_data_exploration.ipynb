### 4.7 Correlation Analysis

Correlation analysis is used to examine the relationships between
numerical features in the student performance datasets.

## 5. Outlier Analysis

Outlier analysis is performed to identify unusual observations
in numerical features using the Interquartile Range (IQR) method.

```python
def detect_outliers(df, columns):
    results = []

    for column in columns:
        Q1 = df[column].quantile(0.25)
        Q3 = df[column].quantile(0.75)
        IQR = Q3 - Q1

        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        outlier_count = (
            (df[column] < lower_bound) |
            (df[column] > upper_bound)
        ).sum()

        results.append({
            "Feature": column,
            "Lower Bound": lower_bound,
            "Upper Bound": upper_bound,
            "Outlier Count": outlier_count
        })

    return pd.DataFrame(results)


numeric_features = [
    "age", "absences", "studytime",
    "failures", "G1", "G2", "G3"
]

print("MATHEMATICS OUTLIERS")
display(detect_outliers(math_df, numeric_features))

print("PORTUGUESE OUTLIERS")
display(detect_outliers(portuguese_df, numeric_features))
```
